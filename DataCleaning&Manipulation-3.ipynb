## Comprobación de Cálculos

```python
# Lo que estoy haciendo es añadir 2 columnas a df calculadas para cada fila
# expected_total: estoy calculando yo el valor esperado según la formula
# amount_difference: veo la diferencia real con el valor registrado para identificar discrepancias

df["expected_total"] = (
    df["quantity"]
    * df["unit_price_eur"]
    * (1 - df["discount_pct"] / 100)
    + df["shipping_cost_eur"]
)

df["amount_difference"] = (
    df["total_amount_eur"] - df["expected_total"]
)

# Ahora vamos a visualizar esto, selecciono solo las columnas que quiero comparar
df[["total_amount_eur", "expected_total", "amount_difference"]].head(10)
# Al comparar si que veo inconsistencias

df.loc[
    df["amount_difference"].abs() > 0.02,
    ["transaction_id", "total_amount_eur",
     "expected_total", "amount_difference"]
]
```

```python
# Cantidades negativas
df[df["quantity"] < 0]
```

```python
# Descuentos superiores a 100
df[df["discount_pct"] > 100]
```

```python
# Fuera del rango: menor que 0 O mayor que 100
df[
    (df["discount_pct"] < 0)
    | (df["discount_pct"] > 100)
]
```

```python
# Venta Y cantidad negativa
df[
    (df["transaction_type"] == "sale")
    & (df["quantity"] < 0)
]
```

## GroupBy

groupby sirve para resumir datos por grupos. Lo utilizas cuando te piden algo como:
- ¿Cuánto se ha vendido por ciudad?
- ¿Cuál es el precio medio por categoría?
- ¿Cuántas transacciones hay por cliente?
La diferencia con lo que ya sabes es: filtrar selecciona filas; groupby agrupa filas y calcula un resumen de cada grupo.

Estructura general: df.groupby("columna_para_agrupar")["columna_para_calcular"].operacion()

### Operaciones
- .sum(): sumar los valores de cada grupo.
- .mean(): calcular la media de cada grupo.
- .median(): calcular la mediana de cada grupo.
- .min(): obtener el valor mínimo de cada grupo.
- .max(): obtener el valor máximo de cada grupo.
- .size(): contar las filas de cada grupo.
- .count(): contar los valores no nulos de cada grupo.
- .nunique(): contar los valores distintos de cada grupo, excluyendo nulos por defecto.
- .std(): medir cuánto varían los valores dentro de cada grupo.
- .agg(): realizar varios cálculos y reunirlos en una tabla.

```python
# Quiero calcular el importe total de cada ciudad
df.groupby("city")["total_amount_eur"].sum()
# Por ejemplo aqui puedo ver que en maadrid es de donde genero mayor ventas
```

```python
# Importe medio de cada ciudad
df.groupby("city")["total_amount_eur"].mean()
```

```python
# Importe máximo por ciudad
df.groupby("city")["total_amount_eur"].max()
```

```python
# Número de clientes distintos conocidos por ciudad
df.groupby("city")["customer_id"].nunique()
```

```python
ventas = df[df["transaction_type"] == "sale"]

resumen = (
    ventas.groupby("city")["total_amount_eur"]
    .sum()
    .reset_index()
    .sort_values("total_amount_eur", ascending=False)
)
print(resumen)
```
